```python
src = "/Volumes/project1/default/raw-data/Python_Basics_tutorial.dbc"
dst = "/Volumes/project1/default/raw-data/Python_Basics_tutorial.txt"

with open(src, "rb") as f:
    raw = f.read()

# Decode using a common DBC encoding
text = raw.decode("latin-1")

with open(dst, "w", encoding="utf-8") as f:
    f.write(text)

print("Converted to text file")
```
